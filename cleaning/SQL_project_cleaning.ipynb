```python
import pandas as pd
```

```python
file_path = 'C:/Users/Korisnik/Desktop/SQL Project/bet_x.csv'
df = pd.read_csv(file_path, encoding="utf-8", low_memory=False)
```

```python
df_clean = df[[
    "ID",
    "Season",
    "Matchday",
    "Match",
    "Date",
    "Time",
    "Home Team",
    "Away Team",
    "Full-Time Home Goals",
    "Full-Time Away Goals",
    "Result",
    "Outcome",
    "My Odds",
    "1xBet Odds",
    "William Hill Odds"
]]
```

```python
df_clean.columns = [
    "id",
    "season",
    "matchday",
    "match",
    "match_date",
    "match_time",
    "home_team",
    "away_team",
    "ft_home_goals",
    "ft_away_goals",
    "result",
    "outcome",
    "my_odds",
    "odds_1xbet",
    "odds_williamhill"
]
```

```python
df_clean["match_date"] = pd.to_datetime(df_clean["match_date"], dayfirst=True)
```

```python
df_clean["ft_home_goals"] = pd.to_numeric(df_clean["ft_home_goals"], errors="coerce")
df_clean["ft_away_goals"] = pd.to_numeric(df_clean["ft_away_goals"], errors="coerce")
df_clean["outcome"] = pd.to_numeric(df_clean["outcome"], errors="coerce")
```

```python
df_clean["matchday"] = df_clean["matchday"].str.extract(r'(\d+)')
df_clean["matchday"] = pd.to_numeric(df_clean["matchday"], errors="coerce")
```

```python
df_clean.to_csv("clean_bet_x.csv", index=False, encoding="utf-8")
```
